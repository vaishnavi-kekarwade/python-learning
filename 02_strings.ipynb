```python
str1 ="This is a string.\nwe are creating in python" # to create strings 
str2 ='discipline beings \nwhen comfort ends'
str3 ="""this is a string"""
#escape sequence characters
#basic operation, concatenation, length of string
print(str1)
print(str2)
```

```python
#basic operation, concatenation, length of string
str1 ="move on "
print(len(str1))
str2 ="carry on"
print(len(str2))
final_str= str1 +" "+str2
print(final_str)
print(len(final_str))
```

```python
#indexing 
str ="Discipline"
print(str[4])
print(str[0])
```

```python
#slicing is accessing parts of a string
str ="Discipline"
print(str[0:4])
print(str[6:len(str)])
print(str[:10]) #[0:10]
print(str[0:])
```

```python
#negative index
str ="management"
print(str[-10:-4])
```

```python
#Store your full name as a string. Print only your first name using slicing.
str = "Vaishnavi Kekarwade"
print(str[0:9])

#Print the last character of your name using negative indexing
print(str[-11:-9])
```

```python
#Join your city and state with a comma using concatenation.
str1 ="Nagpur"
str2 = "Maharashtra"
final_str = str1 +","+str2
print(final_str)
```

```python
#Print your name and college on two separate lines using \n — in one print statement only.
str1 = 'Vaishnavi Kekarwade\nInstitute of Science'
print(str1)
```
