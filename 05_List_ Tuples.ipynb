# List vs Tuple in Python

This notebook explains the difference between Lists and Tuples in Python with examples and use cases.

## Introduction

In Python, Lists and Tuples are used to store multiple items in a single variable.

- Lists are mutable (changeable)
- Tuples are immutable (unchangeable)

## Lists in Python

Lists are ordered, mutable collections.

```python
# Creating a list
subjects = ["Mathematics", "Computers", "Statistics"]

# Accessing elements
print(subjects[2])

# Modifying list
subjects[1] = "Artificial Intelligence"
print(subjects)

# Adding elements
subjects.append("Probability")
print(subjects)
```

```python
marks1 = 94.9
marks2 = 30.5
marks3 = 30.6
marks4 = 50.8
marks5 = 45.9
marks =[94.4,30.5,30,6,50.8,45.9] # we can store multiple type of data in single list
print(marks)
print(type(marks))
print(marks [0])
print(marks[2])
```

```python
#reverse
list = ['a','b','c','d','e','f']
list.reverse()
print(list)
```

```python
#insert  is very similar to append, but it will go to a particular index
#list.insert(index,el)  insert element at index
list=[2,1,3,4]
list.insert(1,2)
print(list)
```

```python
#remove
list =[2,1,22,21]
list.remove(22) # removes first occurreence of element
print(list)
list.pop(1) # removes element at idx
print(list)
```

## Tuples in Python

Tuples are ordered, immutable collections.

```python
# Creating a tuple
numbers = (1983, 1969, 2004) ## numbers[1] = 1983

# Accessing elements
print(numbers[0]) # Trying to modify (will give error)
```

```python
#example
tup=(11,7,5,3)
print(type(tup))
print(tup[0])
print(tup[2])
print(tup[0:3])
```

```python
#tuple methods
tup =(2,1,3,1)
tup.index(3)
print(tup.index(3))
print(tup.count(1))
```

## Difference Between List and Tuple

| Feature       | List            | Tuple           |
|--------------|----------------|----------------|
| Mutability   | Mutable        | Immutable      |
| Syntax       | [ ]            | ( )            |
| Performance  | Slower         | Faster         |
| Use case     | Dynamic data   | Fixed data     |

```python
# Example: Student data

marks = [85, 90, 78] ## List (can change marks)
marks[0] = 88


student = ("VaishNavi", 21, "MCA") ## Tuple (fixed student info)

print(marks)
print(student)
```

```python
#Convert list to tuple
my_list = [1, 2, 3]
my_tuple = tuple(my_list)
print(my_tuple)

#Count occurrences in list
nums = [1, 2, 2, 3, 2]
print(nums.count(2))
```

```python
#write a program to check if aa list contains a palindrome of elements 
list1 = [1,2,3,2,1]
list2 = [1,3,11,4,1]

copy_list1 = list1.copy()
copy_list1.reverse()

if(copy_list1 == list1):
    print("palindrome")
else:
    print("not palindrome")

copy_list2 = list2.copy()
copy_list2.reverse()

if(copy_list2 == list2):
    print("palindrome")
else:
    print("not palindrome")


```
