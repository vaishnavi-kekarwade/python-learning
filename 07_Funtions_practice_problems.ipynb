```python
def count_vowels(word):
    count = 0

    for letter in word:
        if letter in "aeiouAEIOU":
            count += 1

    return count

result = count_vowels("python")
print(result)
```

1. Fibonacci Series
sequence 0 1 1 2 3 5 8
rule : each number = previous two numbers added

logic starts with a = 0 b =1

 for forward nenxt = a+b

```python
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")

        temp = a
        a = b
        b = temp + b

fibonacci(7)
```

Prime number checker
prime number is divisible only by 1 and itself

```python
def is_prime(num):

    if num <=1:
        return False
    for i in range(2, num):

        if num % i == 0:
            return False
        return True
    return True
print(is_prime(7))
```
