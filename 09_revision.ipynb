4. Remove duplicates from a list

```python
#convert the list to a set , sets only stores unique elements so it automatically remove duplicates.

numbers = [11,5,24,1,7,6,4,15,15,5,11]
unique_numbers = list(set(numbers))
print(unique_numbers)
```

5. Count Word Frequency in a Sentence

```python
# split the sentence into words and use the dictionary to keep tarck of the count
sentence = "when there is no enemy within the enemy outside can do us no harm"
words = sentence.split()
frequency ={}

for word in words:
  frequency[word] = frequency.get(word,0) + 1

print(frequency)
```

Type 1 . Count vowels (loop version)

```python
name = "vaishnavi"
count = 0

for letter in name:
    if letter in "aeiou":
        count += 1
print(count)
```

Type 2. Remove duplicates with a loop 


```python
numbers = [11, 4, 5, 24, 7, 9,4,11]
unique =[]

for n in numbers:
    if n not in unique:
        unique.append(n)
print(unique)
```

Type 3. Multiplication table of 8 

```python
num = 8 

for i in range (1,11): # why range (1,11) range stops one number before the end , if u miss the last line , 
    print(num ,"x",i,"=",num * i)
```
