1. Count vowels in your name

```python
name= "vaishnavi"
vowels= "aeiou"
vowel_count = sum(1 for char in name if char in vowels)
print(vowel_count)
```

2. Reverse a string without [::-1]


```python
# we will using a loop or the reversed()  function to invert string manually

text = "ambitious"
reversed_text ="".join(reversed(text))

print(reversed_text)
```

4. Remove duplicates from a list

```python
#convert the list to a set , sets only stores unique elements so it automatically remove duplicates.

numbers = [11,5,24,1,7,6,4,15,15,5,11]
unique_numbers = list(set(numbers))
print(unique_numbers)
```

5. Count Word Frequency in a Sentence

```python
# split the sentence into words and use the dictionary to keep tarck of the count
sentence = "when there is no enemy within the enemy outside can do us no harm"
words = sentence.split()
frequency ={}

for word in words:
  frequency[word] = frequency.get(word,0) + 1

print(frequency)
```
