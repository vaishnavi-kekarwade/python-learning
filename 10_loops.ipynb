Adds all numbers from 1 to 100 using a for loop

```python
total = 0
for i in range(1, 101):   # range(1, 101) gives 1 to 100
    total = total + i
print(total)
```

Prints the multiplication table of a number from 1 to 10.

```python
n = 5
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

Prints a triangle of stars, one more star on each row

```python
for i in range(1, 6):
    print("*" * i)
```
