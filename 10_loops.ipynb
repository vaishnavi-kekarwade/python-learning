Part B - Number problems

#Count even numbers from 1 to 50


```python
count = 0 # counting from zero

for num in range(1,51): # 1 to 50 numbers ,it stops before 51.
    if num % 2 == 0: # even numbers are divisible by 2
        count +=1 # increase the counter by 1.
print(count)      # to print
```

#Reverse a number

```python
# let 123 becomes 321.
# we need to extract the last digit repeatedly and build a reversed number

num =123
reverse = 0

while num > 0:
    digit = num%10  #extracts the last digit
    reverse = reverse * 10 + digit
    num = num // 10 # it will remove the last digit

print(reverse)
```

#Sum of digits

```python
#123 becoes 1+2+3+4 =10
num = 1234
total = 0

while num > 0 :
    digit = num % 100
    total = total + digit 
    num = num // 10

print(total)
```

#Factorial of 5

```python
#factorial means multiplying a positive integer by all positive intergers below it 
# 5! = 5*4*3*2*1=120

num = 5
factorial = 1 # 1 is the multiplication identity , starting with zero make the answer zero

for i in range (1, num+1):
    factorial = factorial * i
print(factorial)
```

Part C- Pattern problems

Pattern problems teach you how nested loops work.
A nested loop means one loop inside another loop.
Right triangle of numbers

```python
for i in range(1,6):
    for j in range(1, i+1): # prints on the same line
        print(j,end="") # moves to the next line
    print()
```

Square of stars (4 × 4)

```python
for i in range(4): # output loop runs 4 times (4 rows)
    for j in range(4): # inner loops runs 4 times
        print("*" , end=" ")
    print()
```

Upside-down triangle / inverted triangle

```python
for i in range(5,0,-1):  # start at 5 , start before 0 ,decrese by 1 each time
    for j in range(i):
        print("*", end=" ")
    print()
```

Part C - List problem

Find the largest number without using max()

```python
numbers = [10,25,7,40,15] # assume the first element is the largest

largest = numbers[0] # compare it with every other element , if u find a bigger number , update the largest variable

for num in numbers: 
    if num > largest:
        largest = num
print(largest) # print the final largest number
```
