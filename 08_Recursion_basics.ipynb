Recursion

The process in which a function calls itself directly or indirectly is called recursion and the corresponding function is called a recursive function.

```python
# Simple example - countdown
def countdown(n):
    if n == 0:     # base case 
        print("Done!")
        return
    print(n)
    countdown(n-1)  # calls itself

countdown(5)
# prints: 5 4 3 2 1 Done!
```

```python
def show(n):
    if (n==-1): # base case
        return
    print(n)
    show(n-1)
    print("END")
show(4) #5,4,3,2,1 4 = n-3
# use recursion
```

Two rules of recursion:

Always have a base case (stopping condition)
Each call must move towards the base case

Write a recursive function for Factorial using recursion

```python
#factorial
def fact(n):
    if (n ==1 or n==0):
        return 1
    return fact(n-1)*n
print(fact(4))
```

Write a recursive function to calculate the sum of first n natural numbers

```python
def calc_sum(n):
    if(n==0):
        return 0
    #print(n)
    return calc_sum(n-1)+ n
sum = calc_sum(6)
print(sum)
```

Fibonacci using Recursion

```python
def fibonacci(n):
    if n <=1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

#using example
n = 6

for i in range(n):
    print(fibonacci(i), end=" ")
```

Sum of Digits using Recursion

```python
def sum_of_digits(n):
# base condition

  if n ==0: # recursion keeps calling the function till n ==0 that is our base case.
    return 0
  
  #logic
# n% 10 --- gives last digits
# n// 10 --- removes last digits
  return n % 10  + sum_of_digits(n//10)

#example
num = 1234 # 1234 % 10 = 4 , 123 // 10 = 123
print("sum of digits:", sum_of_digits(num))
```

Reverse String using Recursion

```python
def reverse_string(s):
    # first base condition
    if len(s) ==0:
        return s
    
    #logic:
    #s[1:] -- remaininig string
    #s[0] -- first character
    return reverse_string(s[1:]) + s[0]

#example

text= " study hard"
print("reversed string:", reverse_string(text))
```

Power Function using Recursion
Used to calculate power of a number recursively.

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base,exponent-1)

print(power(2,3))

# 2* power(2,2)
# 2 * 2 * power(2,1)
# 2*2*2 * power(2,0) ;; 2*2*2*1 = 8
```

Print Numbers 1 to N using Recursion

```python
def print_1_to_n(n):
    #base condition

    if n == 0: # function first goes till 0 .
        return
    # recursive call

    print_1_to_n(n-1)

    # print while returning back
    print(n)

print_1_to_n(7)
```

Print Numbers N to 1 using Recursion

```python
def print_n_to_1(n):
    # base condition

    if n == 0:
        return
    # print first
    print(n)


    # recursive call

    print_n_to_1(n-1)

print_n_to_1(5)
```

Sum of First N Numbers using Recursion

```python
def sum_n(n):

    # base condition

    if n == 0:
        return 0
    
    # recursive formula

    return n + sum_n(n-1)

print(sum_n(5))
```

Count Digits using Recursion

```python
def count_digits(n):

    #base condition

    if n == 0:
        return 0
    
    # remove one digit and count

    return 1 + count_digits(n // 10)

print(count_digits(12345))
```

Product of Digits using Recursion

```python
def product_digits(n):

    #base condition

    if n < 10:
        return n
    
    #multiply last digit ith remaining number

    return (n % 10) * product_digits(n//10) # n% 10 gives last digit , n // 10 removes last digit
print(product_digits(1234))
```

GCD/HCF


what is gcd / hcf
GCD (Greatest Common Divisor) or HCF (Highest Common Factor) means:

Largest number that divides both numbers completely.

GCD of 12 and 18 = 6 --->6 divides both 12 and 18

```python
def gcd(a,b):

    #base condition
    if b == 0:
        return a
    
    # recursive call
    return gcd(b,a%b) # uses euclidean alogrithm

print(gcd(12,18))

#gcd(12, 18)

#18 % 12 = 6

→ #gcd(18, 12)
```

Binary Search using Recursion

Understanding the Logic

Binary Search works only on a sorted list.

Instead of checking every element one by one:
It checks the middle element
Then removes half of the search space

Example Array
Index:    0   1   2   3   4   5
Array:   [10, 20, 30, 40, 50, 60]

We want to find:

target = 40

Function Parameters:
binary_search(arr, low, high, target)

```python
def binary_search(arr, low  , high , target)# function parameter
    
    #base conditon

    if low > high:
        return -1
    
    # find middle index

    mid =(low +high) //2 # find middle

    # if a target found

    if arr[mid] == target:  # compare target
        return mid
    
    #search in left half

    elif target < arr[mid]: # decide direction
        return binary_search(arr,low,mid -1,target)
    
    # search in right half

    else:
        return binary_search(arr,mid +1,high,target) # recursive call
    
    #example

    numbers = [10,20,30,40,50,60]

    result = binary_search(numbers,0,len(numbers) -1 , 40)

    print("element found at index:" , result)
```

1. Sorted Array Required

Binary Search works only on sorted arrays.
Every step removes half of the array.

Binary Search works by:

Finding middle element
Comparing target
Searching only one half
Repeating recursively until found
